Student Performance

The dataset is related to student performance and attributes in secondary education.Here's a general introduction based on the columns present in the DataFrame

This dataset contains information about students in secondary education,including demographic details,family background,educational support,lifestyle choices, and academic performance.Each row represents a unique student,and the columns provide various attributes and characteristics associated with them.

Dataset contains the following columns:

1.school:School attended by the student

2.sex:Student's gender(male or female)

3.address:Type of address of the the student(urban or rural)

4.famsize:Family size(binary categorization:'LE3'- less or equal to 3 or 'GT3'-greater than 3)

5.Pstatus:Parent's cohabitation status(living together or apart)

6.Medu:Mother's education level(0 to 4,representing none to higher education)

7.Fedu:Father's education level(0 to 4,representing none to higher education)

8.Mjob:Mother's occupation

8.Fjob:Father's occupation

9.reason:Reason for choosing the school

10.guardian:Student's guardian

11.traveltime:Travel time to school

12.studytime:Weekly study time(hours)

13.failures:Number of past class failures

14.schoolsup:Extra educational support at school(yes or no)

15.famsup:Family educational support(yes or no)

16.paid:Extra paid classes within the course subject(yes or no)

17.activities:Extra-curricular activities(yes or no)

18.nursery:Attended nursery school(yes or no)

19.higher:Wants to take higher education(yes or no)

20.internet:internet access at home(yes or no)

21.romantic:In a romantic relationship(yes or no)

22.famrel:Quality of family relationship(from 1- very bad to 5 - excellent)

23.freetime:Free time after school(from 1-very low to 5 - very high)

24.goout:Going out with friends(from 1-very low to 5 -very high)

25.Dalc:Workday alcohol consumption(from 1-very low to 5-very high)

26.Walc:Weekend alcohol consumption(from 1-vey low to 5-vey high)

27.health:Current health status(from 1-very bad to 5 - very good)

28.absences: Number of school absences

29.G1:First period grade(from 0 to 20)

29.G2:Second period grade(from 0 to 20)

29.G3:Final grade grade(from 0 to 20)

### # Outlier detection


```python

students_grades_df=df.copy()
students_grades_df.drop(['G1','G2','G3'],axis=1)
```

```python
# Function to detect outliers

def detect_outliers(columns):
    outlier_indices=[]
    for column in columns:
        Q1=students_grades_df[column].quantile(0.25)
        Q3=students_grades_df[column].quantile(0.75)
        IQR=Q3-Q1

        # mask is boolean condition(true or false)
        mask=(students_grades_df[column]>=(Q1-1.5)*IQR) & (students_grades_df[column]<=(Q1+1.5)*IQR)

        # Bool value to numpy array
        mask=mask.to_numpy()

        # Find the indices of values that do not satisfy the mask condition --> that are outliers
        false_indices=np.argwhere(~mask)
        outlier_indices.append(false_indices)

    return np.unique(np.concatenate(outlier_indices).ravel())





```

```python
num_cols=['age','absences']
detect_outliers(num_cols)
```

## Model Training

```python
# Data split
X,y=students_grades_df.iloc[:,:-1],students_grades_df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
```

```python
# Identifying categorical and numerical columns
categorical_features = X.select_dtypes(include=['object']).columns
numerical_features = X.select_dtypes(exclude=['object']).columns

print(f"Categorical features: {list(categorical_features)}")
print(f"Numerical features: {list(numerical_features)}")
```

```python
# Creating a column transformer to one-hot encode categorical features and pass through numerical

preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# Splitting the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
```

#### Model Training and Comparison

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Defining the models to train
models = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(),
    'Lasso Regression': Lasso(),
    'Decision Tree': DecisionTreeRegressor(random_state=42),
    'Random Forest': RandomForestRegressor(random_state=42),
    'Gradient Boosting': GradientBoostingRegressor(random_state=42)
}

# Dictionary to store results
results = {}

# Looping through each model, train, and evaluate
for name, model in models.items():
    pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                               ('model', model)])
    pipeline.fit(X_train, y_train)
    y_pred = pipeline.predict(X_test)
    
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    
    results[name] = {'MAE': mae, 'MSE': mse, 'R2': r2}
    print(f"{name}")
    print(f"  MAE: {mae:.4f}")
    print(f"  MSE: {mse:.4f}")
    print(f"  R2 : {r2:.4f}\n")

# Comparing results
results_df = pd.DataFrame(results).T
print("Model Comparison (Sorted by R2):")
print(results_df.sort_values(by='R2', ascending=False))
```

#### Hyperparameter Tuning for the Best Model

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
import numpy as np

lasso_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', Lasso(random_state=42, max_iter=10000))
])

param_grid = {
    'model__alpha': [0.0001, 0.0005, 0.001, 0.005, 0.01, 0.05, 
                     0.1, 0.5, 1.0, 5.0, 10.0]
}

grid_search = GridSearchCV(
    lasso_pipeline,
    param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)

# Fit the grid search
grid_search.fit(X_train, y_train)

print("\n")
print(f"Best parameters found: {grid_search.best_params_}")
print(f"Best cross-validation R² score: {grid_search.best_score_:.4f}")
```

```python
# Fine-Grained Search Around the Best Alpha

# Getting the best alpha from the coarse search
best_alpha_coarse = grid_search.best_params_['model__alpha']
print(f"Coarse best alpha: {best_alpha_coarse}")

fine_alphas = np.linspace(max(0.00001, best_alpha_coarse * 0.1), 
                          best_alpha_coarse * 2, 20)

param_grid_fine = {'model__alpha': fine_alphas}

lasso_pipeline_fine = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', Lasso(random_state=42, max_iter=10000))
])

grid_search_fine = GridSearchCV(
    lasso_pipeline_fine,
    param_grid_fine,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)

grid_search_fine.fit(X_train, y_train)

print(f"\nBest alpha (fine): {grid_search_fine.best_params_['model__alpha']:.6f}")
print(f"Best CV R²       : {grid_search_fine.best_score_:.4f}")
```

####  Evaluate the Tuned Model

```python

best_lasso = grid_search_fine.best_estimator_
y_pred_tuned = best_lasso.predict(X_test)

mae_tuned = mean_absolute_error(y_test, y_pred_tuned)
mse_tuned = mean_squared_error(y_test, y_pred_tuned)
r2_tuned = r2_score(y_test, y_pred_tuned)

TOLERANCE = 1
accuracy_tuned = np.mean(np.abs(y_test - y_pred_tuned) <= TOLERANCE)

print("TUNED LASSO REGRESSION")
print(f"Best Alpha        : {grid_search_fine.best_params_['model__alpha']:.6f}")
print(f"R² Score          : {r2_tuned:.4f}")
print(f"MAE               : {mae_tuned:.4f}")
print(f"MSE               : {mse_tuned:.4f}")
print(f"Accuracy (±{TOLERANCE} mark): {accuracy_tuned:.2%}")
print("\n")

# Comparing to baseline Lasso
print("IMPROVEMENT OVER BASELINE LASSO (R²=0.8823, MAE=0.6773, MSE=1.1869)")
print(f"R²  change: {r2_tuned - 0.8823:+.4f}")
print(f"MAE change: {mae_tuned - 0.6773:+.4f}")
print(f"MSE change: {mse_tuned - 1.1869:+.4f}")
```

#### Comparision Table

```python


# Prediction counts as "correct" if within ±1 mark of actual G3
TOLERANCE = 1  

results = {}
all_predictions = {} 

for name, model in models.items():
    pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                               ('model', model)])
    pipeline.fit(X_train, y_train)
    y_pred = pipeline.predict(X_test)
    all_predictions[name] = y_pred
    
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    
    # Tolerance-based accuracy: fraction of predictions within ±TOLERANCE of actual
    accuracy = np.mean(np.abs(y_test - y_pred) <= TOLERANCE)
    
    results[name] = {
        'MAE': mae,
        'MSE': mse,
        'R2': r2,
        f'Accuracy (±{TOLERANCE})': accuracy
    }

results_df = pd.DataFrame(results).T
results_df_sorted = results_df.sort_values(by='R2', ascending=False)

print(f"Model Comparison (with tolerance-based Accuracy, ±{TOLERANCE} mark):")
print(results_df_sorted.round(4))
```

```python
# Set up the figure with 4 subplots
fig, axes = plt.subplots(1, 4, figsize=(26, 6))
fig.suptitle(f'Model Benchmark Comparison (Accuracy defined as Prediction within ±{TOLERANCE} mark)',
             fontsize=18, fontweight='bold', y=1.03)

accuracy_col = f'Accuracy (±{TOLERANCE})'

# Plot 1: MAE (lower is better) 
sns.barplot(x=results_df_sorted.index, y=results_df_sorted['MAE'],
            ax=axes[0], palette='Reds_r')
axes[0].set_title('Mean Absolute Error\n(Lower is Better)', fontsize=12)
axes[0].set_ylabel('MAE')
axes[0].set_xlabel('')
axes[0].tick_params(axis='x', rotation=45, labelsize=9)
for i, v in enumerate(results_df_sorted['MAE']):
    axes[0].text(i, v + 0.02, f'{v:.3f}', ha='center', fontsize=9, fontweight='bold')

# Plot 2: MSE (lower is better)
sns.barplot(x=results_df_sorted.index, y=results_df_sorted['MSE'],
            ax=axes[1], palette='Oranges_r')
axes[1].set_title('Mean Squared Error\n(Lower is Better)', fontsize=12)
axes[1].set_ylabel('MSE')
axes[1].set_xlabel('')
axes[1].tick_params(axis='x', rotation=45, labelsize=9)
for i, v in enumerate(results_df_sorted['MSE']):
    axes[1].text(i, v + 0.05, f'{v:.3f}', ha='center', fontsize=9, fontweight='bold')

# Plot 3: R2 (higher is better)
sns.barplot(x=results_df_sorted.index, y=results_df_sorted['R2'],
            ax=axes[2], palette='Greens')
axes[2].set_title('R² Score\n(Higher is Better)', fontsize=12)
axes[2].set_ylabel('R² Score')
axes[2].set_xlabel('')
axes[2].tick_params(axis='x', rotation=45, labelsize=9)
axes[2].axhline(y=0, color='red', linestyle='--', linewidth=1, alpha=0.6)
for i, v in enumerate(results_df_sorted['R2']):
    axes[2].text(i, v + 0.01, f'{v:.3f}', ha='center', fontsize=9, fontweight='bold')

# Plot 4: Accuracy (higher is better)
sns.barplot(x=results_df_sorted.index, y=results_df_sorted[accuracy_col],
            ax=axes[3], palette='Blues')
axes[3].set_title(f'Accuracy (±{TOLERANCE} mark)\n(Higher is Better)', fontsize=12)
axes[3].set_ylabel('Accuracy')
axes[3].set_xlabel('')
axes[3].tick_params(axis='x', rotation=45, labelsize=9)
axes[3].set_ylim(0, 1.05)
for i, v in enumerate(results_df_sorted[accuracy_col]):
    axes[3].text(i, v + 0.015, f'{v:.1%}', ha='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

####  Accuracy at Multiple Tolerances

```python
# Defining multiple tolerance thresholds
tolerances = [0.5, 1, 1.5, 2, 2.5, 3]

# Computing accuracy for each model at each tolerance level
accuracy_curves = {}

for name in models.keys():
    y_pred = all_predictions[name]
    accuracies = [np.mean(np.abs(y_test - y_pred) <= t) for t in tolerances]
    accuracy_curves[name] = accuracies

accuracy_df = pd.DataFrame(accuracy_curves, index=[f'±{t}' for t in tolerances])

# Plot
plt.figure(figsize=(12, 6))
markers = ['o', 's', '^', 'D', 'v', 'P']

for idx, (model_name, values) in enumerate(accuracy_curves.items()):
    plt.plot(tolerances, values, marker=markers[idx], linewidth=2, 
             markersize=8, label=model_name)

plt.xlabel('Tolerance (marks)', fontsize=12)
plt.ylabel('Accuracy', fontsize=12)
plt.title('Model Accuracy vs. Tolerance Threshold', fontsize=14, fontweight='bold')
plt.xticks(tolerances)
plt.ylim(0, 1.05)
plt.axhline(y=0.5, color='grey', linestyle=':', alpha=0.5, label='50% Baseline')
plt.grid(alpha=0.3)
plt.legend(loc='lower right', fontsize=10)
plt.tight_layout()
plt.show()

# Print the summary
print("\nAccuracy at Each Tolerance Level:")
print((accuracy_df * 100).round(1).astype(str) + '%')
```

### Saving the model

```python
import joblib
import os
import json
import numpy as np

os.makedirs('student_performance_model', exist_ok=True)

# Save the pipeline
joblib.dump(best_lasso, 'student_performance_model/lasso_pipeline.pkl')

# Build metadata with everything the app needs (NO dependency on X at runtime)
metadata = {
    'best_alpha': float(grid_search_fine.best_params_['model__alpha']),
    'test_r2': float(r2_tuned),
    'test_mae': float(mae_tuned),
    'test_mse': float(mse_tuned),
    'test_accuracy_tolerance_1': float(accuracy_tuned),
    'categorical_features': list(categorical_features),
    'numerical_features': list(numerical_features),
    'all_features': list(X.columns),
}

# Categorical valid values
for col in categorical_features:
    metadata[f'{col}_values'] = sorted(X[col].unique().tolist())

# Numerical ranges AND medians (medians used as default values in the app)
for col in numerical_features:
    metadata[f'{col}_range'] = [float(X[col].min()), float(X[col].max())]
    metadata[f'{col}_median'] = float(X[col].median())

metadata['G3_range'] = [int(y.min()), int(y.max())]

with open('student_performance_model/metadata.json', 'w') as f:
    json.dump(metadata, f, indent=2)

```
